```python
import random

# Global comparison counter
comparison_count = 0


def min_max_second(arr, low, high):
    global comparison_count

    # Base Case: One Element
    if low == high:
        return arr[low], None, arr[low], None

    # Base Case: Two Elements
    if high == low + 1:
        comparison_count += 1

        if arr[low] < arr[high]:
            return arr[low], arr[high], arr[high], arr[low]
        else:
            return arr[high], arr[low], arr[low], arr[high]

    # Divide
    mid = (low + high) // 2

    lmin, l2min, lmax, l2max = min_max_second(arr, low, mid)
    rmin, r2min, rmax, r2max = min_max_second(arr, mid + 1, high)

    # ---------- Combine Minimum ----------
    comparison_count += 1
    if lmin < rmin:
        overall_min = lmin

        if l2min is None:
            second_min = rmin
        elif rmin < l2min:
            comparison_count += 1
            second_min = rmin
        else:
            comparison_count += 1
            second_min = l2min
    else:
        overall_min = rmin

        if r2min is None:
            second_min = lmin
        elif lmin < r2min:
            comparison_count += 1
            second_min = lmin
        else:
            comparison_count += 1
            second_min = r2min

    # ---------- Combine Maximum ----------
    comparison_count += 1
    if lmax > rmax:
        overall_max = lmax

        if l2max is None:
            second_max = rmax
        elif rmax > l2max:
            comparison_count += 1
            second_max = rmax
        else:
            comparison_count += 1
            second_max = l2max
    else:
        overall_max = rmax

        if r2max is None:
            second_max = lmax
        elif lmax > r2max:
            comparison_count += 1
            second_max = lmax
        else:
            comparison_count += 1
            second_max = r2max

    return overall_min, second_min, overall_max, second_max


# -------------------------
# Example
# -------------------------

arr = [15, 3, 27, 9, 40, 6, 12, 35, 2, 18]

comparison_count = 0

minimum, second_minimum, maximum, second_maximum = min_max_second(
    arr, 0, len(arr) - 1
)

print("Array :", arr)
print("Minimum :", minimum)
print("Second Minimum :", second_minimum)
print("Maximum :", maximum)
print("Second Maximum :", second_maximum)
print("Comparisons :", comparison_count)
```
