```python
import heapq

# ---------------------------------
# Modified Dijkstra's Algorithm
# (Distance + Hop Count)
# ---------------------------------
def dijkstra(graph, source):
    n = len(graph)

    distance = [float('inf')] * n
    hops = [float('inf')] * n
    previous = [None] * n

    distance[source] = 0
    hops[source] = 0

    # (distance, hops, vertex)
    pq = [(0, 0, source)]

    while pq:
        dist, hop, u = heapq.heappop(pq)

        if dist > distance[u]:
            continue

        for v, weight in graph[u]:
            new_dist = distance[u] + weight
            new_hops = hops[u] + 1

            # Better distance OR same distance but fewer hops
            if (new_dist < distance[v]) or \
               (new_dist == distance[v] and new_hops < hops[v]):

                distance[v] = new_dist
                hops[v] = new_hops
                previous[v] = u

                heapq.heappush(pq, (new_dist, new_hops, v))

    return distance, hops, previous


# ---------------------------------
# Path Reconstruction
# ---------------------------------
def get_path(previous, source, destination):
    path = []

    while destination is not None:
        path.append(destination)
        destination = previous[destination]

    path.reverse()

    if path[0] == source:
        return path
    return []


# ---------------------------------
# Graph with 6 Vertices
# ---------------------------------
graph = {
    0: [(1, 2), (2, 1)],
    1: [(5, 3)],
    2: [(3, 1)],
    3: [(5, 3)],
    4: [(5, 2)],
    5: []
}

source = 0

distance, hops, previous = dijkstra(graph, source)

print("Modified Dijkstra's Algorithm")
print("-" * 65)
print(f'{"Vertex":>8} {"Distance":>10} {"Hops":>8} {"Path":>25}')
print("-" * 65)

for v in range(len(graph)):
    path = get_path(previous, source, v)
    path_str = " -> ".join(map(str, path)) if path else "No Path"

    d = distance[v] if distance[v] != float('inf') else "INF"
    h = hops[v] if hops[v] != float('inf') else "-"

    print(f"{v:>8} {str(d):>10} {str(h):>8} {path_str:>25}")
```
