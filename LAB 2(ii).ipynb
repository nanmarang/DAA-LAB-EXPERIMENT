```python
def compute_lps(pattern):
    """Compute the Longest Prefix Suffix (LPS) array."""
    m = len(pattern)
    lps = [0] * m

    length = 0
    i = 1

    print("Building LPS Array")
    print("-" * 40)

    while i < m:
        if pattern[i] == pattern[length]:
            length += 1
            lps[i] = length
            print(f"Match: pattern[{i}] = '{pattern[i]}' -> LPS[{i}] = {length}")
            i += 1
        else:
            if length != 0:
                print(f"Mismatch at index {i}, fallback length = {lps[length-1]}")
                length = lps[length - 1]
            else:
                lps[i] = 0
                print(f"Mismatch: LPS[{i}] = 0")
                i += 1

    print("\nFinal LPS Array:", lps)
    return lps


def kmp_trace(text, pattern):
    """KMP search with complete tracing."""
    n = len(text)
    m = len(pattern)

    lps = compute_lps(pattern)

    print("\n")
    print("=" * 60)
    print("KMP SEARCH TRACE")
    print("=" * 60)

    i = 0
    j = 0
    shifts = 0
    comparisons = 0
    matches = []

    while i < n:
        comparisons += 1

        print(f"\nComparing Text[{i}]='{text[i]}' with Pattern[{j}]='{pattern[j]}'")

        if text[i] == pattern[j]:
            print("Match")
            i += 1
            j += 1

            if j == m:
                print(f"\nPattern Found at Index {i - j}")
                matches.append(i - j)

                shifts += 1
                print(f"Shift Pattern using LPS[{j-1}] = {lps[j-1]}")

                j = lps[j - 1]

        else:
            print("Mismatch")

            if j != 0:
                print(f"Shift Pattern -> j = LPS[{j-1}] = {lps[j-1]}")
                j = lps[j - 1]
                shifts += 1
            else:
                print("Move Text Pointer")
                i += 1
                shifts += 1

    print("\n")
    print("=" * 60)
    print("RESULT")
    print("=" * 60)
    print("Matches :", matches)
    print("Comparisons :", comparisons)
    print("Total Shifts :", shifts)


# ---------------- MAIN ----------------

text = "AABCAABXAAAZ"
pattern = "AABXAA"

print("Text    :", text)
print("Pattern :", pattern)
print()

kmp_trace(text, pattern)
```
