```python
import random
import time
import heapq

INF = float('inf')

# ---------------------------------
# Generate Dense Graph (100 Vertices)
# ---------------------------------
V = 100
random.seed(10)

# Adjacency Matrix
matrix = [[0] * V for _ in range(V)]

for i in range(V):
    for j in range(i + 1, V):
        weight = random.randint(1, 100)
        matrix[i][j] = weight
        matrix[j][i] = weight

# Convert Matrix to Adjacency List
adj = {i: [] for i in range(V)}

for i in range(V):
    for j in range(V):
        if matrix[i][j] != 0:
            adj[i].append((j, matrix[i][j]))


# ---------------------------------
# Prim's Algorithm (Adjacency Matrix)
# Time Complexity: O(V²)
# ---------------------------------
def prim_matrix(graph):
    n = len(graph)

    selected = [False] * n
    key = [INF] * n
    parent = [-1] * n

    key[0] = 0

    total_cost = 0

    for _ in range(n):

        u = -1
        minimum = INF

        for v in range(n):
            if not selected[v] and key[v] < minimum:
                minimum = key[v]
                u = v

        selected[u] = True
        total_cost += key[u]

        for v in range(n):
            if graph[u][v] != 0 and not selected[v]:
                if graph[u][v] < key[v]:
                    key[v] = graph[u][v]
                    parent[v] = u

    return total_cost


# ---------------------------------
# Prim's Algorithm (Heap)
# Time Complexity: O((V+E) logV)
# ---------------------------------
def prim_heap(adj):
    visited = [False] * V

    pq = [(0, 0)]

    total_cost = 0

    while pq:

        weight, u = heapq.heappop(pq)

        if visited[u]:
            continue

        visited[u] = True
        total_cost += weight

        for v, w in adj[u]:
            if not visited[v]:
                heapq.heappush(pq, (w, v))

    return total_cost


# ---------------------------------
# Performance Comparison
# ---------------------------------

start = time.perf_counter()
cost_matrix = prim_matrix(matrix)
matrix_time = time.perf_counter() - start

start = time.perf_counter()
cost_heap = prim_heap(adj)
heap_time = time.perf_counter() - start


print("Prim using Adjacency Matrix")
print("---------------------------")
print("MST Cost :", cost_matrix)
print("Execution Time :", matrix_time, "seconds")

print()

print("Prim using Heap")
print("---------------------------")
print("MST Cost :", cost_heap)
print("Execution Time :", heap_time, "seconds")
```
