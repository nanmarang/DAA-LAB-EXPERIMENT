```python
import random

# -----------------------------
# Naive 2D Pattern Matching
# -----------------------------
def naive_2d_pattern_match(text, pattern):
    rows_text = len(text)
    cols_text = len(text[0])

    rows_pattern = len(pattern)
    cols_pattern = len(pattern[0])

    matches = []
    comparisons = 0

    # Slide the pattern over the text matrix
    for i in range(rows_text - rows_pattern + 1):
        for j in range(cols_text - cols_pattern + 1):

            match = True

            # Compare every element
            for x in range(rows_pattern):
                for y in range(cols_pattern):
                    comparisons += 1

                    if text[i + x][j + y] != pattern[x][y]:
                        match = False
                        break

                if not match:
                    break

            if match:
                matches.append((i, j))

    return matches, comparisons


# -----------------------------
# Generate a 10x10 Text Matrix
# -----------------------------
random.seed(10)

text = [[random.randint(0, 9) for _ in range(10)] for _ in range(10)]

# Create a known 3x3 pattern
pattern = [
    [4, 5, 6],
    [7, 8, 9],
    [1, 2, 3]
]

# Insert the pattern into the text at position (5,4)
insert_row = 5
insert_col = 4

for i in range(3):
    for j in range(3):
        text[insert_row + i][insert_col + j] = pattern[i][j]

# -----------------------------
# Display Text Matrix
# -----------------------------
print("10 x 10 TEXT MATRIX")
print("-" * 35)
for row in text:
    print(row)

print("\n3 x 3 PATTERN MATRIX")
print("-" * 20)
for row in pattern:
    print(row)

# -----------------------------
# Search Pattern
# -----------------------------
matches, comparisons = naive_2d_pattern_match(text, pattern)

print("\nSearch Results")
print("-" * 35)

if matches:
    for r, c in matches:
        print(f"Pattern found at Row {r}, Column {c}")
else:
    print("Pattern not found.")

print(f"\nTotal Element Comparisons = {comparisons}")
```
