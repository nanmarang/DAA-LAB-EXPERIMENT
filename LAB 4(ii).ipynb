```python
import heapq

# ---------------------------------
# Dijkstra's Algorithm
# ---------------------------------
def dijkstra(graph, source):
    n = len(graph)

    distance = [float('inf')] * n
    previous = [None] * n

    distance[source] = 0

    pq = [(0, source)]

    while pq:
        current_dist, u = heapq.heappop(pq)

        if current_dist > distance[u]:
            continue

        for v, weight in graph[u]:
            new_dist = distance[u] + weight

            if new_dist < distance[v]:
                distance[v] = new_dist
                previous[v] = u
                heapq.heappush(pq, (new_dist, v))

    return distance, previous


# ---------------------------------
# Reconstruct Shortest Path
# ---------------------------------
def shortest_path(previous, source, destination):
    path = []

    current = destination

    while current is not None:
        path.append(current)
        current = previous[current]

    path.reverse()

    if path[0] == source:
        return path
    return []


# ---------------------------------
# Directed Graph (City Road Network)
# ---------------------------------
graph = {
    0: [(1, 10), (2, 3)],       # Airport
    1: [(2, 1), (3, 2)],        # City Center
    2: [(1, 4), (3, 8), (4, 2)],# Bus Terminal
    3: [(4, 7)],                # Shopping Mall
    4: [(3, 9)]                 # Central Station
}

source = 0      # Airport
destination = 4 # Central Station

distance, previous = dijkstra(graph, source)

path = shortest_path(previous, source, destination)

print("===== Dijkstra's Algorithm =====")
print("Source      : Airport (Node 0)")
print("Destination : Central Station (Node 4)\n")

print("Fastest Route:")
print(" -> ".join(map(str, path)))

print("\nTravel Time =", distance[destination], "minutes")
```
